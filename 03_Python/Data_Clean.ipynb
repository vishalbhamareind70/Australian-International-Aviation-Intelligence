# 2. Data Cleaning

## 2.1 Correct Passenger Data Types

### Objective
Convert passenger count columns from text (`object`) to numeric data types for accurate calculations and analysis.

### Method
Use Pandas to convert passenger values into numbers and check for any missing values introduced during conversion.


```python

# Select passenger columns that need numeric conversion
passenger_cols = ["PaxIn", "PaxOut", "TotalPax"]
airline_cols = ["Passengers In", "Passengers Out"]

# Convert passenger columns from text to numbers
passenger_df[passenger_cols] = passenger_df[passenger_cols].apply(
    pd.to_numeric, errors="coerce"
)

airline_df[airline_cols] = airline_df[airline_cols].apply(
    pd.to_numeric, errors="coerce"
)

# Check the data types and missing values after conversion
for name, df, cols in [
    ("International City Pair", passenger_df, passenger_cols),
    ("Airlines By Country", airline_df, airline_cols)
]:
    print(f"\n{name}")
    print(pd.DataFrame({
        "Data Type": df[cols].dtypes,
        "Missing Values": df[cols].isna().sum()
    }))
```

```python

# Reload original passenger data for investigation
original_passenger = pd.read_excel(
    DATA_DIR / "International City Pair.xlsx",
    sheet_name="Data", header=0
)

original_airline = pd.read_excel(
    DATA_DIR / "Airlines By Country.xlsx",
    sheet_name="Data", header=0
)

# Check values that cannot be converted directly to numbers
for name, df, columns in [
    ("International City Pair", original_passenger, passenger_cols),
    ("Airlines By Country", original_airline, airline_cols)
]:
    print(f"\n{name}")

    for col in columns:
        invalid = df.loc[
            pd.to_numeric(df[col], errors="coerce").isna()
            & df[col].notna(),
            col
        ]

        print(f"{col}: {invalid.value_counts().head(10).to_dict()}")
```


### Observations

- Passenger count columns were successfully converted to numeric data types.
- Non-numeric values were identified as `..` placeholders.
- These placeholders were converted to `NaN` to preserve missing information without introducing artificial passenger counts.
- Missing passenger values will be considered when performing calculations and aggregations.



## 2.2 Handle Missing Values

### Findings
- The passenger datasets contained `..` placeholders, which were converted to `NaN` using `pd.to_numeric(errors="coerce")`.
- The Flights and Seats dataset already contained missing values recognised by Pandas as `NaN`.
- `NaN` represents missing or unavailable data, not necessarily zero.

### Cleaning Decision
- Retain missing values as `NaN` rather than replacing them with zero.
- Avoid deleting records unnecessarily to preserve historical information.
- Handle missing values according to the requirements of each analysis.

This approach prevents artificial values from affecting passenger trends, flight capacity calculations, and forecasting.



## 2.3 Investigate Duplicate Route-Month Records

### Objective
Examine the four repeated route-month combinations identified in the Flights and Seats dataset.

### Method
Display the affected records to determine whether they contain identical or different flight and seat capacity values.


```python

# Columns that identify a route in a particular month
route_keys = ["Month", "Australian", "International"]

# Find all records belonging to repeated route-month combinations
duplicate_routes = capacity_df[
    capacity_df.duplicated(subset=route_keys, keep=False)
].sort_values(route_keys)

# Display the records for investigation
print("Affected records:", len(duplicate_routes))
print("Repeated route-month combinations:",
      duplicate_routes[route_keys].drop_duplicates().shape[0])

display(duplicate_routes)
```


### Observations

- Four repeated route-month combinations were identified, producing eight affected records.
- All occur on the Sydney–Beijing route between March and June 2026.
- The repeated records contain different flight and seat values and are not exact duplicates.
- No records were deleted. These combinations will require appropriate treatment before route-level aggregation or merging.



## 2.4 Standardise Column Names

### Objective
Rename columns using a consistent and readable naming convention.

### Method
Use lowercase `snake_case` names (words separated by underscores) while preserving the original data and meaning of each column.


```python

# Rename columns in the City Pair dataset
passenger_df = passenger_df.rename(columns={
    "Month": "month",
    "AustralianPort": "australian_port",
    "ForeignPort": "foreign_port",
    "Country": "country",
    "PaxIn": "passengers_in",
    "PaxOut": "passengers_out",
    "TotalPax": "total_passengers"
})

# Rename columns in the Flights and Seats dataset
capacity_df = capacity_df.rename(columns={
    "Month": "month",
    "Australian": "australian_port",
    "International": "foreign_port",
    "(a) InFlightsExc": "in_flights_exclusive",
    "(b) InFlightsAll": "in_flights_all",
    "(a) OutFlightsExc": "out_flights_exclusive",
    "(b) OutFlightsAll": "out_flights_all",
    "(a) InSeatsMin": "in_seats_min",
    "(b) InSeatsMax": "in_seats_max",
    "(a) OutSeatsMin": "out_seats_min",
    "(b) OutSeatsMax": "out_seats_max"
})

# Rename columns in the Airlines By Country dataset
airline_df = airline_df.rename(columns={
    "Month": "month",
    "Scheduled Operator": "airline",
    "Country to/from": "country",
    "Passengers In": "passengers_in",
    "Passengers Out": "passengers_out"
})

# Verify updated column names
for name, df in [
    ("International City Pair", passenger_df),
    ("International Flights and Seats", capacity_df),
    ("Airlines By Country", airline_df)
]:
    print(f"\n{name}")
    print(df.columns.tolist())
```


## 2.5 Standardise Categorical Values

### Objective
Ensure consistent formatting of airport, country, and airline names across the three datasets.

### Method
Remove unnecessary surrounding spaces and convert text values to uppercase to support accurate grouping, filtering, and dataset merging.


```python

# Define text columns in each dataset
text_columns = {
    "International City Pair": (
        passenger_df, ["australian_port", "foreign_port", "country"]
    ),
    "International Flights and Seats": (
        capacity_df, ["australian_port", "foreign_port"]
    ),
    "Airlines By Country": (
        airline_df, ["airline", "country"]
    )
}

# Standardise text values across all three datasets
for name, (df, columns) in text_columns.items():
    for col in columns:
        df[col] = df[col].str.strip().str.upper()

    print(f"{name}: Text columns standardised successfully")
```


## 2.6 Check Missing Months in Historical Data

### Objective
Identify missing calendar months to ensure the datasets have continuous historical coverage.

### Method
Generate a complete monthly date range and compare it with the available months in each dataset.

### Importance
Missing months can affect trend analysis, visualisations, and forecasting. Any gaps will be identified and documented rather than automatically filled with zero.


```python

# Check missing months across all datasets
for name, df in [
    ("International City Pair", passenger_df),
    ("International Flights and Seats", capacity_df),
    ("Airlines By Country", airline_df)
]:
    # Create a complete monthly timeline
    expected_months = pd.date_range(
        df["month"].min(),
        df["month"].max(),
        freq="MS"
    )

    # Find months absent from the dataset
    missing_months = expected_months.difference(df["month"].unique())

    print(f"\n{name}")
    print("Missing Months:", len(missing_months))
    print("Dates:", missing_months.strftime("%Y-%m").tolist())
```


### Observations

- International City Pair and Airlines By Country have complete monthly coverage from January 2009 to June 2026.
- International Flights and Seats contains 20 missing calendar months.
- Eighteen missing months occur between 2003 and 2005, suggesting a different reporting frequency during the early period.
- September and October 2022 are also missing from the Flights and Seats dataset.
- Missing months will not be artificially filled. Historical comparisons will use available and compatible reporting periods.
